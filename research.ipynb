# Описание данных

Описание данных. 
	airports_nearest — расстояние до ближайшего аэропорта в метрах (м).   
	balcony — число балконов.   
	ceiling_height — высота потолков (м).   
	cityCenters_nearest — расстояние до центра города (м).   
	days_exposition — сколько дней было размещено объявление (от публикации до снятия).   
	first_day_exposition — дата публикации.   
	floor — этаж.   
	floors_total — всего этажей в доме.   
	is_apartment — апартаменты (булев тип).   
	kitchen_area — площадь кухни в квадратных метрах (м²).   
	last_price — цена на момент снятия с публикации.   
	living_area — жилая площадь в квадратных метрах (м²).   
	locality_name — название населённого пункта.   
	open_plan — свободная планировка (булев тип).   
	parks_around3000 — число парков в радиусе 3 км.   
	parks_nearest — расстояние до ближайшего парка (м).   
	ponds_around3000 — число водоёмов в радиусе 3 км.   
	ponds_nearest — расстояние до ближайшего водоёма (м).   
	rooms — число комнат.   
	studio — квартира-студия (булев тип).   
	total_area — общая площадь квартиры в квадратных метрах (м²).   
	total_images — число фотографий квартиры в объявлении.

## `floors_total` и `locality_name`

Строки, для который доля пропуска в столбцах составляет < 5%, я предлагаю удалить (`floors_total`,`locality_name`) Для столбцов, где доля пропусков составляет > 10% простое удаление строк не подходит, можно потерять много информации. Обработаем каждую колонку индивидуально.

```python
df = df.dropna(subset=['floors_total'])
df = df.dropna(subset=['locality_name'])
```

```python
df['floors_total'] = df['floors_total'].astype(int)
```

Так как `locality_name` - расположение жилья, а расположение прямо вляет на цену, лучше удалить малое количество строк с пропусками и не загрязнять данные значением моды (для категориального признака). 

Также в `floors_total` есть значение в 60 этажей, по данным в поиске в интернете: самое высокое здание в г. Санкт-Петербург имеет 37 этажей, возьму это значение как планку

```python
df = df.drop(index=df.loc[df['floors_total'] > 37].index)
```

## `ceiling_height`

Следующий признак - `ceiling_height` (высота потолков). В столбце есть выбросы.

```python
df.boxplot(column='ceiling_height')
```

Видно наглядный выброс, в виде значения 100, его удалим. Остальные "выбросы" могут означать, например, что 34 метра это 3,4 метра, просто по ошибке не поставили запятую. Обработаем такие значения. Значения от 10 метров до 24 удалим, так как это аномалия и связать с ошибкой сложно. 

```python
index_to_drop = df[(df['ceiling_height'] > 60) | ((df['ceiling_height'] > 10) & (df['ceiling_height'] < 25))].index
df = df.drop(index_to_drop)
```

```python
to_change = (df['ceiling_height'] >= 25) & (df['ceiling_height'] < 60)
df.loc[to_change,'ceiling_height'] = df.loc[to_change,'ceiling_height'] / 10
```

Так же по рисунку сверху видно, что у нижней границы есть значения, близкие к 0, например, 1,2 метра не пригодная для жизни высота потолков. Удалим значения меньше 2,5 метра (это минимум строительных норм)

```python
df['ceiling_height'].nsmallest(40)
```

```python
df = df.drop(index=df.loc[df['ceiling_height'] < 2.5].index)
```

```python
df['ceiling_height'].hist(bins=50)
```

```python
print(df['ceiling_height'].mean())
print(df['ceiling_height'].median())
```

```python
df.boxplot(column='ceiling_height')
```

Поменяем пропущенные значения на медиану потому что распределение неравномерное

```python
df['ceiling_height'] = df['ceiling_height'].fillna(df['ceiling_height'].median())
```
